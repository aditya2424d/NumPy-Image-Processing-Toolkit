# NumPy Image Processing Toolkit
Treat an image as a NumPy array: grayscale, flip, rotate, split channels, build a collage, and compare vectorized operations with pixel loops.


```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data

img = data.astronaut()
print("Shape:", img.shape)
print("Data type:", img.dtype)

plt.figure(figsize=(6, 6))
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

## 1. Grayscale Conversion
Use the RGB channels directly as NumPy arrays and compute grayscale values with vectorized arithmetic.


```python
r = img[:, :, 0]
g = img[:, :, 1]
b = img[:, :, 2]

gray = (0.299 * r + 0.587 * g + 0.114 * b).astype(np.uint8)

plt.figure(figsize=(6, 6))
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image")
plt.axis("off")
plt.show()
```

## 2. Flip and Rotate
Flip with slicing and rotate 90° using transpose plus slicing.


```python
flipped = img[:, ::-1, :]
rotated = np.transpose(img, (1, 0, 2))[::-1, :, :]

plt.figure(figsize=(15, 5))
for i, (image, title) in enumerate([(img, "Original"), (flipped, "Flipped"), (rotated, "Rotated 90°")], 1):
    plt.subplot(1, 3, i)
    plt.imshow(image)
    plt.title(title)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

## 3. Split RGB Channels


```python
red = np.zeros_like(img)
green = np.zeros_like(img)
blue = np.zeros_like(img)

red[:, :, 0] = img[:, :, 0]
green[:, :, 1] = img[:, :, 1]
blue[:, :, 2] = img[:, :, 2]

plt.figure(figsize=(15, 5))
for i, (image, title) in enumerate([(red, "Red Channel"), (green, "Green Channel"), (blue, "Blue Channel")], 1):
    plt.subplot(1, 3, i)
    plt.imshow(image)
    plt.title(title)
    plt.axis("off")
plt.tight_layout()
plt.show()
```

## 4. Build a Collage
Stack images horizontally and vertically. All four images have matching dimensions.


```python
top = np.hstack((img, flipped))
bottom = np.hstack((rotated, img))
collage = np.vstack((top, bottom))

plt.figure(figsize=(10, 10))
plt.imshow(collage)
plt.title("NumPy Image Collage")
plt.axis("off")
plt.show()
```

## 5. Vectorized NumPy vs Pixel Loop
The two functions perform the same grayscale calculation. `%timeit` shows the performance difference.


```python
def numpy_gray(image):
    return (
        0.299 * image[:, :, 0]
        + 0.587 * image[:, :, 1]
        + 0.114 * image[:, :, 2]
    )

def loop_gray(image):
    h, w, _ = image.shape
    output = np.zeros((h, w))
    for i in range(h):
        for j in range(w):
            r, g, b = image[i, j]
            output[i, j] = 0.299 * r + 0.587 * g + 0.114 * b
    return output
```

```python
%timeit numpy_gray(img)
```

```python
%timeit loop_gray(img)
```
